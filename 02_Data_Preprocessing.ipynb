#### Import libraries

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import re
import string
import warnings
warnings.filterwarnings('ignore')
```

```python
# Downloading NLTK data
nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('punkt_tab', quiet=True)

print("All libraries imported successfully!")
```

### 1. Load the dataset

```python
# Load the dataset
df = pd.read_csv("bbc_news.csv")


print(f"Dataset loaded: {df.shape}")
print(f"Columns: {df.columns.tolist()}")

df.head()
```

### 2. Preprocessing Steps

##### Step 1: Lowercasing

```python
df['text_lower'] = df['text'].str.lower()

print("Text converted to lowercase")
print("\nExample:")
print("Original:", df['text'].iloc[0][:100])
print("Lowercase:", df['text_lower'].iloc[0][:100])
```

##### Step 2: Remove Punctuation, Symbols, and Numbers

```python
def remove_punctuation_numbers(text):
    # Remove numbers
    text = re.sub(r'\d+', '', text)
    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    return text

df['text_clean'] = df['text_lower'].apply(remove_punctuation_numbers)

print("Punctuation and numbers removed")
print("\nExample:")
print("Before:", df['text_lower'].iloc[0][:100])
print("After:", df['text_clean'].iloc[0][:100])
```

##### Step 3: Tokenization

```python
df['tokens'] = df['text_clean'].apply(word_tokenize)

print("Text tokenized")
print("\nExample:")
print("Text:", df['text_clean'].iloc[0][:100])
print("Tokens:", df['tokens'].iloc[0][:20])
```

##### Step 4: Stop-word Removal

```python
stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]

df['tokens_no_stop'] = df['tokens'].apply(remove_stopwords)

print("Stopwords removed")
print(f"\nTotal stopwords: {len(stop_words)}")
print("\nExample:")
print("Before:", df['tokens'].iloc[0][:20])
print("\nAfter:", df['tokens_no_stop'].iloc[0][:20])
```

##### Step 5: Lemmatization

```python
lemmatizer = WordNetLemmatizer()

def lemmatize_tokens(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

df['tokens_lemmatized'] = df['tokens_no_stop'].apply(lemmatize_tokens)

print("Tokens lemmatized")
print("\nExample:")
print("Before:", df['tokens_no_stop'].iloc[0][:20])
print("\nAfter:", df['tokens_lemmatized'].iloc[0][:20])
```

##### Step 6: Join Tokens Back to Text

```python
df['processed_text'] = df['tokens_lemmatized'].apply(lambda x: ' '.join(x))

print("Tokens joined back to text")
print("\nExample:")
print("Processed text:", df['processed_text'].iloc[0][:200])
```

### 4. Before and After Comparison

```python
print("\n" + "="*70)
print("PREPROCESSING COMPARISON")
print("="*70)

sample_idx = 0
print(f"\nCategory: {df['label'].iloc[sample_idx]}")
print("\nORIGINAL TEXT:")
print(df['text'].iloc[sample_idx][:300])
print("\nPROCESSED TEXT:")
print(df['processed_text'].iloc[sample_idx][:300])
```

```python
# Statistics
print("\n" + "="*70)
print("STATISTICS:")
print("="*70)
print(f"Average original length: {df['text'].str.len().mean():.0f} characters")
print(f"Average processed length: {df['processed_text'].str.len().mean():.0f} characters")
```

### 5. Save Preprocessed Data

```python
# Save preprocessed data
df_final = df[['text', 'processed_text', 'label']].copy()
df_final.to_csv('bbc_news_preprocessed.csv', index=False)

print("Preprocessed data saved as bbc_news_preprocessed.csv")
print("Shape:", df_final.shape)
```
